# IF-ELIF-ELSE 

```python
if a >= b and a >= c:
    print(a)
elif b >= a and b >= c:
    print(b)
else:
    print(c)
```

```python
#Calculator

first = int(input("Enter your first number:"))
second = int(input("Enter your second number:"))
operator = input("Enter your operator'+,-,*,/':")
if operator == "+":
    print("Addition of two numbers:",first+second)
elif operator == "-":
    print("Subtraction of two numbers:",first - second)
elif operator == "*":
    print("Multiplication of two numbers:",first*second)
elif operator == "/":
    print("Division if two numbers:",first/second)
else:
    print("Invalid Operator")
```

```python
# Discount offers
Purchase_amount = int(input("Enter your amount to know discount offer:"))
if Purchase_amount >= 20000:
    print(f"You get the 20% discount for your amount:{Purchase_amount}")
elif Purchase_amount >= 10000:
    print(f"You get the 10% discount for your amount:{Purchase_amount}")
elif Purchase_amount >= 5000:
    print(f"You get the 5% discount for your amount:{Purchase_amount}")
else:
    print("To get the discount buy minimum 5000 of the product")
```

# Nested IF

```python
#ATM Code
card_visiblity = True
correct_pin = 1234
balance = 200000
if card_visiblity == True:
    enter_pin = int(input("Enter you pin:"))
    if enter_pin == correct_pin:
        credit = int(input("Enter your withdraw amount:"))
        if balance >= credit:
            print("Please collect your money")
            print("Your account balance is:",balance - credit)
        else:
            print("There is no enough balance in account")
    else:
        print("Please enter correct pin")
else:
    print("Insert your card")
    
```
