## Mini Project

```python
student = {}
```

```python
def add_std(std_id, name, marks):
    student[std_id] = {"name":name,"marks":marks}
    print(name, "Added Successfully!!")
```

```python
def view_std():
    if len(student)==0:
        print("No students")
        return
    for std_id,details in student.items():
        print("ID:",std_id,"--","Name:",details["name"],"Marks:",details["marks"])
```

```python
def search_std(std_id):
    if std_id in student:
        details = student[std_id]
        print("FOUND",details["name"], "Marks",details["marks"])
    else:
        print("NOT FOUND by your id",std_id)
```

```python
def del_std(std_id):
    if std_id in student:
        del student[std_id]
        print("Deleted Successfully!")
    else:
        print(std_id,"ID NOT FOUND")
```

```python
def show_menu():
    print("""
    ###################
    |1.Add Student     |
    |2.View Student    | 
    |3.Search Student  |
    |4.Delete Student  |
    |5.Exit            |
    ###################
    """
         )
```

```python
def run_program():
    while True:
        show_menu()
        choice = input("Enter your Choice:")

        if choice == "1":
            std_id = input("Enter your Id:")
            name = input("Enter Name:")
            marks = int(input("Enter your marks"))
            add_std(std_id,name, marks)
        elif choice == "2":
            view_std()
        elif choice == "3":
            std_id = input("Enter your id to search:")
            search_std(std_id)
        elif choice == "4":
            std_id = input("Enter your id to delete:")
            del_std(std_id)
        elif choice == "5":
            print("Exited Successfully")
            break
        else:
            print("Invalid choice")
```

```python
run_program()
```
