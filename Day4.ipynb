```python
def division(x,y):
    try:
        res = x / y
    except TypeError:
        print("You should Enter both Integers only not strings")
    except ZeroDivisionError:
        print("Nunber is not divided by 0")
    else:
        print("Division Successfully done",res)
    finally:
        print("Division Done successfully",x ,"/", y, "=", res)
```

```python
division(2,10)
```

```python
division(2,0)
```

```python
division("a","b")
```

# Method with parameters

```python
class Shopping_Cart:
    def __init__(self):
        self.total = 0
    def add_item(self, price,quantity):
        self.price = price
        self.quantity = quantity
        self.total += price * quantity
    def remove_item(self,price,quantity):
        self.total -= price * quantity
```

```python
cart = Shopping_Cart()
```

```python
cart.add_item(200,3)
```

```python
cart.total
```

```python
cart.add_item(50,1)
```

```python
cart.total
```

```python
cart.add_item(350,1)
```

```python
cart.total
```

```python
cart.remove_item(100,2)
```

```python
cart.total
```

# Mini project

```python
#Game between human and computer
import random as r
human_score = 0
computer_score = 0
print("======================================  WELCOME TO GUESS NUMBER GAME  ===================================================\n")
print("Guess number from 1 to 15\n")
print("Press 'Q' to exit from game\n")

while True:
    user = input("Enter your guess from(1 to 15) or Q to quit:")
    if user.upper() == "Q":
        break
    elif not user.isdigit():
        print("Please enter digit from(1 to 15)\n")
        continue
    human_guess = int(user)
    if human_guess < 0 and human_guess > 15:
        print("Please enter number b/t(1 to 15):\n")
        continue
    computer_guess = r.randint(1,15)
    print(f"Computer guess: {computer_guess}")
    print(f"Human guess: {human_guess}")
    if human_guess == computer_guess:
        human_score += 1
        print("Correct Human get 1 point")
    else:
        computer_score += 1
        print("Sorry human, Computer won this game",computer_score)
    print(f"Computer Score:{computer_score} || Human Score:{human_score}")
    print("=" * 20)
print("======= Final Score ========")
print(f"Human Score:{human_score}")
print(f"Computer Score:{computer_score}")
if human_score > computer_score:
    print("Human Won the game🏆")
elif computer_score > human_score:
    print("Computer Won the game🏆")
else:
    print("It is Draw between human and Computer")
    print("If you want to win try game again until you win")
    
```
